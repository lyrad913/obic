```python
from pathlib import Path
import polars as pl
import os

ARTIFACTS_DIR = Path("../artifacts")
MODEL_FILES = [
    "bigger_lgbm_3.csv",
    "bigger_lgbm_seed_1337.csv",
    "bigger_lgbm_seed_2025.csv",
]
OUTPUT_FILE = ARTIFACTS_DIR / "ensemble.csv"

# 각 csv 파일 읽기
dataframes = [pl.read_csv(os.path.join(ARTIFACTS_DIR, file)) for file in MODEL_FILES]
```

```python
dataframes[0]
```

```python
# 같은 행(time, pv_id)의 nins 컬럼끼리 평균 내기
ensemble_df = dataframes[0].with_columns(
    ((dataframes[0]['nins'] + dataframes[1]['nins'] + dataframes[2]['nins']) / 3).alias('nins')
).select(['time', 'pv_id', 'type', 'nins'])

ensemble_df.write_csv(OUTPUT_FILE)
```
